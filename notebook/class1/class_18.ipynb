# class 19 Tools

langchain 提供了工具的定义和执行功能,一般步骤：

1. 用`tool`类。
2. 定义工具的名称，描述，参数等。
3. 实现工具的执行逻辑。

定义一个天气的tool, 注意里面的 3 处定义，示例: 

```python
from langchain.tools import tool

@tool
def get_weather(location: str, weather: str = "晴朗"):
    """
    获取指定位置的天气信息。
    Args:
        location: 位置名称，例如"北京"。
        weather: 天气描述，默认值为"晴朗"。
    Returns:
        天气信息，例如"天气晴朗，温度22摄氏度。"
    """
    return f"天气{weather}, 适合游玩"
```
